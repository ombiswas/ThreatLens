This notebook profiles class distributions, feature relationships, missing/infinite values, and correlation patterns.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loading import load_dataset
from src.preprocessing import clean_dataframe

# Load data
df, path = load_dataset()
print(f"Loaded {len(df):,} records from {path.name}")
```

```python
# Clean and map classes
cleaned_df, label_col, before, after = clean_dataframe(df)
print("Cleaned DataFrame Shape:", cleaned_df.shape)

# Class distribution
plt.figure(figsize=(7, 4))
sns.countplot(data=cleaned_df, x=label_col, palette="viridis")
plt.title("Target Class Distribution")
plt.ylabel("Flow Record Count")
plt.show()
```

```python
# Numeric distributions and summary
print(cleaned_df.describe().T[['mean', 'std', 'min', '50%', 'max']].head(10))
```
